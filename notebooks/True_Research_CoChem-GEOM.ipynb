```python
import logging
import os
import subprocess
import sys
from pathlib import Path

logging.basicConfig(level=logging.INFO, format="%(message)s")
logger = logging.getLogger(__name__)

cwd = Path.cwd().resolve()
module_dir = cwd.parent if cwd.name == "notebooks" else cwd
tests_dir = module_dir / "tests"
if not tests_dir.is_dir():
    raise FileNotFoundError(f"Expected tests directory at {tests_dir}")

logger.info("[CoChem-GEOM] Native Jupyter Test Execution (%s)", module_dir)
env = os.environ.copy()
src_dir = module_dir / "src"
pythonpath_entries = [str(p) for p in (src_dir, module_dir) if p.is_dir()]
if env.get("PYTHONPATH"):
    pythonpath_entries.append(env["PYTHONPATH"])
env["PYTHONPATH"] = os.pathsep.join(pythonpath_entries)

cmd = [sys.executable, "-B", "-m", "pytest", str(tests_dir), "-v"]
result = subprocess.run(cmd, capture_output=True, text=True, cwd=str(module_dir), env=env)

logger.info(result.stdout)
if result.stderr:
    logger.error(result.stderr)
if result.returncode != 0:
    raise RuntimeError(f"Physical execution failed for CoChem-GEOM (exit code {result.returncode}).")
logger.info("Genuine Native Execution Complete.")
```
